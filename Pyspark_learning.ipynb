####date_format()


it changes the date format but returns a string as an output 

formats:
dd-MM-yyyy
yyyy-MM-dd

note the MM is in uppercase else it give incorrect values 

String → Date : to_date()

Date → String : date_format()

The one that ends in a date type is to_date; 

anything named *_format gives you back text.

```python
df=df.drop('week_before')
```

```python
df=df.withColumn('week_before',date_add('curr_date',-7))
df.display()
```

```python
df=df.withColumn('week_before',date_format('week_before','dd-MM-yyyy'))
df.display()
```

Your input strings were in dd-MM-yyyy format (like 18-07-2026), and to_date parsed them correctly into DateType. But here's the key point: once a value is a DateType, Spark always displays it as yyyy-MM-dd — that's the internal/standard representation. You can't change how a DateType displays; the dd-MM-yyyy in your code only describes the input format, not the output.
So 2026-07-18 is correct — that's 18-07-2026 stored as a proper date.
If you actually want the column shown as dd-MM-yyyy, you have to convert it back to a string with date_format — but then it's a string again, not a date


```python
df=df.withColumn('week_before',to_date('week_before','dd-MM-yyyy'))
df.display()
```

**dropping Nulls**

when we want to remove rows which has null values in **all** the columns

```python
df.dropna('all').display()
```

**Filling Nulls**

.fillna()

```python
df.fillna('Not Available').display()
```

when we want to fill null values with **same** value in **all** **columns** specified in **subset()**

```python
df.fillna('Not available',subset=['Outlet_Size','Item_Weight']).display()
```
