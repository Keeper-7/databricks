####date_format()


it changes the date format but returns a string as an output 

formats:
dd-MM-yyyy
yyyy-MM-dd

note the MM is in uppercase else it give incorrect values 

String → Date : to_date()

Date → String : date_format()

The one that ends in a date type is to_date; 

anything named *_format gives you back text.

```python
df=df.drop('week_before')
```

```python
df=df.withColumn('week_before',date_add('curr_date',-7))
df.display()
```

```python
df=df.withColumn('week_before',date_format('week_before','dd-MM-yyyy'))
df.display()
```

Your input strings were in dd-MM-yyyy format (like 18-07-2026), and to_date parsed them correctly into DateType. But here's the key point: once a value is a DateType, Spark always displays it as yyyy-MM-dd — that's the internal/standard representation. You can't change how a DateType displays; the dd-MM-yyyy in your code only describes the input format, not the output.
So 2026-07-18 is correct — that's 18-07-2026 stored as a proper date.
If you actually want the column shown as dd-MM-yyyy, you have to convert it back to a string with date_format — but then it's a string again, not a date


```python
df=df.withColumn('week_before',to_date('week_before','dd-MM-yyyy'))
df.display()
```

##.over()
###.partitionBy()

.over() fuction : what would be the order of the opteration, if we want to genrate over some column 

.over() attaches a window spec to a function, turning it into a window function: it computes a value for each row based on a set of related rows, without collapsing them like groupBy does.

example:

from pyspark.sql import Window
from pyspark.sql.functions import row_number, sum as _sum

w = Window.partitionBy("country").orderBy("year")

df.withColumn("rn", row_number().over(w))

The contrast with groupBy is the point: 
df.**groupBy**("country").sum("sales") **gives you one row per country**. 
_sum("sales").**over**(**Window.partitionBy**("country")) gives you **every original row**, each ****carrying** its country's total alongside it**.

####The three parts of a window spec
w = (Window
     .partitionBy("country")           # split rows into groups
     .orderBy("year")                  # order within each group
     .rowsBetween(-2, 0))              # which rows count for this row

**All** three are **optional**. **partitionBy** **alone** treats each group as one **unordered block**. **orderBy** gives you **position-aware behavior**. The frame narrows things further.


###What you can put in front of .over()
**row_number**()      # 1,2,3,4 — always distinct
**rank**()            # 1,2,2,4 — ties share, gaps after
**dense_rank**()      # 1,2,2,3 — ties share, no gaps
**ntile**(4)          # bucket number
**percent_rank**()

**Aggregates** — sum, avg, min, max, count, collect_list, etc.

**Analytic** — reach across rows:

from pyspark.sql.functions import lag, lead, first, last

df.withColumn("prev_sales", lag("sales", 1).over(w)) \
  .withColumn("delta", col("sales") - col("prev_sales"))


The default frame (this catches people)

The frame changes depending on whether you called orderBy:
w1 = Window.partitionBy("country")
-# frame: entire partition → total per country, same value on every row

w2 = Window.partitionBy("country").orderBy("year")
-# frame: unbounded preceding → current row → running total

So adding an orderBy silently converts a group total into a cumulative sum. If you want the whole-partition total and an ordering, say so explicitly:

w = (Window.partitionBy("country").orderBy("year")
     .rowsBetween(Window.unboundedPreceding, Window.unboundedFollowing))

**A worked example — top N per group**

w = Window.partitionBy("dept").orderBy(col("salary").desc())

top3 = (df.withColumn("rn", row_number().over(w))
          .filter(col("rn") <= 3)
          .drop("rn"))

**APPEND** : add data we want to write to exiting data present at the file location

**OVERWRITE** : overwrite the data present in the file location 

**ERROR**: check if the file path has data then throws ERROR else write the data in the path

**IGNORE**: check if the file path had data then it IGNORE( doesn't throw any error or write the data) else write

APPEND

```python
df.write.format('csv').mode('append').save('/Workspace/Shared/Writing data_1/')
```

Alternate way of writing

```python
df.write.format('csv').mode('append').option('header','true').option('path','/Workspace/Shared/Writing data_1/').save()
```

ERROR

```python
df.write.format('csv').mode('error').option('header','true').option('path','/Workspace/Shared/Writing data_2/').save()
```

IGNORE

```python
df.write.format('csv').mode('ignore').option('header','true').option('path','/Workspace/Shared/Writing data_1/').save()
```

OVERWRITE

```python
df.write.format('csv').mode('overwrite').option('header','true').option('path','/Workspace/Shared/Writing data_1/').save()
```
