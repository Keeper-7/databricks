PySpark Question

You are given a PySpark DataFrame containing three city columns:

| City1 | City2 | City3 |
| ----- | ----- | ----- |
| Goa   | null  | AP    |
| null  | AP    | null  |
| null  | null  | Bglr  |

Write PySpark code to transform the DataFrame into a single-column DataFrame containing all the non-null city values.

**Expected Output**

| Result |
| ------ |
| Goa    |
| AP     |
| Bglr   |

**Requirements**

Ignore all NULL values.

The output should contain only one column named Result.

Try to solve it using PySpark DataFrame APIs, not SQL.

Bonus: Find at least two different approaches to solve the problem.

Smaple input data 

data = [
    ("Goa", None, "AP"),
    (None, "AP", None),
    (None, None, "Bglr")
]

df = spark.createDataFrame(data, ["City1", "City2", "City3"])

df.show()

```python
data = [
    ("Goa", None, "AP"),
    (None, "AP", None),
    (None, None, "Bglr")
]

df_city = spark.createDataFrame(data, ["City1", "City2", "City3"])

df_city.show()
```

```python
df_city_result=df_city.withColumn('Result',coalesce(df_city['City1'],df_city['City2'],df_city['City3'])).select('Result')

df_city_result.display()
```
